```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("fatigue_log.csv")

plt.figure()
plt.plot(df["eye_open"])
plt.title("Eye Openness Over Time")
plt.xlabel("Frame")
plt.ylabel("Eye Openness (pixels)")
plt.show()
```

```python
plt.figure()
plt.plot(df["perclos"])
plt.axhline(0.25, linestyle="--")
plt.axhline(0.40, linestyle="--")
plt.title("PERCLOS Over Time")
plt.xlabel("Frame")
plt.ylabel("PERCLOS")
plt.show()
```

```python
plt.figure()
plt.plot(df["mar"])
plt.title("Mouth Aspect Ratio (Yawning)")
plt.xlabel("Frame")
plt.ylabel("MAR")
plt.show()
```

```python
df["status"].value_counts().plot(kind="bar")
plt.title("Fatigue State Distribution")
plt.show()
```
